```python
opr1=int(input("Enter 1st Operand"))
opr2=int(input("Enter 2nd Operand"))
optr=input("Enter an Arithematic Operator")
if(optr=="+"):
    print("The Add of ",opr1," and ",opr2," is =",(opr1+opr2))
elif(optr=="-"):
        print("The Sub of ",opr1," and ",opr2," is =",(opr1-opr2))
elif(optr=="*"):
        print("The Mul of ",opr1," and ",opr2," is =",(opr1*opr2))
elif(optr=="/"):
        print("The Div of ",opr1," and ",opr2," is =",(opr1/opr2))
else:
    print("Invalid Operator")
```
